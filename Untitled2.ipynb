```python
import os
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2

from pathlib import Path

from skimage.feature import hog
from skimage.feature import local_binary_pattern
from skimage.feature import graycomatrix, graycoprops

from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator

print("TensorFlow version:", tf.__version__)
```

```python
import kagglehub

dataset_path = kagglehub.dataset_download(
    "paultimothymooney/chest-xray-pneumonia"
)

print("Dataset downloaded to:")
print(dataset_path)
```

```python
DATASET_ROOT = dataset_path / "chest_xray"

TRAIN_DIR = DATASET_ROOT / "train"
TEST_DIR = DATASET_ROOT / "test"
VAL_DIR = DATASET_ROOT / "val"

print("Train:", TRAIN_DIR)
print("Validation:", VAL_DIR)
print("Test:", TEST_DIR)
```

```python
classes = ["NORMAL", "PNEUMONIA"]

def count_images(folder):
    counts = {}

    for class_name in classes:
        class_path = folder / class_name

        counts[class_name] = len([
            f for f in class_path.iterdir()
            if f.suffix.lower() in [".jpg", ".jpeg", ".png"]
        ])

    return counts


train_counts = count_images(TRAIN_DIR)
val_counts = count_images(VAL_DIR)
test_counts = count_images(TEST_DIR)

print("Training:", train_counts)
print("Validation:", val_counts)
print("Testing:", test_counts)
```

```python
total_images = (
    sum(train_counts.values()) +
    sum(val_counts.values()) +
    sum(test_counts.values())
)

print("Total images:", total_images)
```

```python
class_balance = pd.DataFrame({
    "Class": classes,
    "Train": [train_counts[c] for c in classes],
    "Validation": [val_counts[c] for c in classes],
    "Test": [test_counts[c] for c in classes]
})

class_balance
```

```python
def display_random_images(folder, class_name, n=8):

    class_path = folder / class_name

    image_files = [
        f for f in class_path.iterdir()
        if f.suffix.lower() in [".jpg", ".jpeg", ".png"]
    ]

    selected_files = random.sample(
        image_files,
        min(n, len(image_files))
    )

    plt.figure(figsize=(16, 8))

    for i, image_file in enumerate(selected_files):

        image = cv2.imread(str(image_file))
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        plt.subplot(2, 4, i + 1)
        plt.imshow(image, cmap="gray")
        plt.title(class_name)
        plt.axis("off")

    plt.tight_layout()
    plt.show()


display_random_images(TRAIN_DIR, "NORMAL", 8)
display_random_images(TRAIN_DIR, "PNEUMONIA", 8)
```

```python
image_paths = []
labels = []

for class_name in classes:

    class_path = TRAIN_DIR / class_name

    for image_file in class_path.iterdir():

        if image_file.suffix.lower() in [".jpg", ".jpeg", ".png"]:

            image_paths.append(str(image_file))

            if class_name == "NORMAL":
                labels.append(0)
            else:
                labels.append(1)


image_paths = np.array(image_paths)
labels = np.array(labels)

print("Total training images:", len(image_paths))
print("Labels shape:", labels.shape)
```

```python
all_image_paths = []
all_labels = []

for split_dir in [TRAIN_DIR, VAL_DIR, TEST_DIR]:

    for class_name in classes:

        class_path = split_dir / class_name

        for image_file in class_path.iterdir():

            if image_file.suffix.lower() in [".jpg", ".jpeg", ".png"]:

                all_image_paths.append(str(image_file))

                if class_name == "NORMAL":
                    all_labels.append(0)
                else:
                    all_labels.append(1)


all_image_paths = np.array(all_image_paths)
all_labels = np.array(all_labels)

print("Total images:", len(all_image_paths))
```

```python
X_train_paths, X_temp_paths, y_train, y_temp = train_test_split(
    all_image_paths,
    all_labels,
    test_size=0.30,
    stratify=all_labels,
    random_state=42
)

X_val_paths, X_test_paths, y_val, y_test = train_test_split(
    X_temp_paths,
    y_temp,
    test_size=0.50,
    stratify=y_temp,
    random_state=42
)

print("Training:", len(X_train_paths))
print("Validation:", len(X_val_paths))
print("Testing:", len(X_test_paths))

print("\nPercentages:")

print("Training:",
      len(X_train_paths) / len(all_image_paths) * 100, "%")

print("Validation:",
      len(X_val_paths) / len(all_image_paths) * 100, "%")

print("Testing:",
      len(X_test_paths) / len(all_image_paths) * 100, "%")
```

```python
IMG_SIZE = 128

def load_and_preprocess_image(path):

    image = cv2.imread(path)

    image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    image = cv2.resize(
        image,
        (IMG_SIZE, IMG_SIZE)
    )

    return image
```

```python
sample_image = load_and_preprocess_image(X_train_paths[0])

plt.figure(figsize=(5, 5))
plt.imshow(sample_image, cmap="gray")
plt.title("Preprocessed Chest X-Ray - 128x128")
plt.axis("off")
plt.show()

print("Image shape:", sample_image.shape)
```

```python
def extract_hog(image):

    features = hog(
        image,
        orientations=9,
        pixels_per_cell=(8, 8),
        cells_per_block=(2, 2),
        block_norm="L2-Hys"
    )

    return features
```

```python
def extract_lbp(image):

    radius = 1
    n_points = 8 * radius

    lbp = local_binary_pattern(
        image,
        n_points,
        radius,
        method="uniform"
    )

    n_bins = n_points + 2

    histogram, _ = np.histogram(
        lbp.ravel(),
        bins=n_bins,
        range=(0, n_bins),
        density=True
    )

    return histogram
```

```python
def extract_glcm(image):

    image_8bit = cv2.normalize(
        image,
        None,
        0,
        7,
        cv2.NORM_MINMAX
    ).astype(np.uint8)

    glcm = graycomatrix(
        image_8bit,
        distances=[1],
        angles=[0],
        levels=8,
        symmetric=True,
        normed=True
    )

    features = [
        graycoprops(glcm, "contrast")[0, 0],
        graycoprops(glcm, "dissimilarity")[0, 0],
        graycoprops(glcm, "homogeneity")[0, 0],
        graycoprops(glcm, "energy")[0, 0],
        graycoprops(glcm, "correlation")[0, 0],
        graycoprops(glcm, "ASM")[0, 0]
    ]

    return np.array(features)
```

```python
def extract_all_features(path):

    image = load_and_preprocess_image(path)

    hog_features = extract_hog(image)
    lbp_features = extract_lbp(image)
    glcm_features = extract_glcm(image)

    combined_features = np.concatenate([
        hog_features,
        lbp_features,
        glcm_features
    ])

    return combined_features
```

```python
features = extract_all_features(X_train_paths[0])

print("Feature vector shape:", features.shape)
print("First 20 features:")
print(features[:20])
```

```python
def create_feature_dataset(image_paths):

    feature_list = []

    for i, path in enumerate(image_paths):

        feature_vector = extract_all_features(path)

        feature_list.append(feature_vector)

        if (i + 1) % 500 == 0:
            print("Processed:", i + 1)

    return np.array(feature_list)
```

```python
X_train_features = create_feature_dataset(X_train_paths)

X_val_features = create_feature_dataset(X_val_paths)

X_test_features = create_feature_dataset(X_test_paths)

print("Training feature shape:", X_train_features.shape)
print("Validation feature shape:", X_val_features.shape)
print("Testing feature shape:", X_test_features.shape)
```

```python
def load_images(paths):

    images = []

    for path in paths:

        image = load_and_preprocess_image(path)

        image = image.astype("float32") / 255.0

        image = np.expand_dims(image, axis=-1)

        images.append(image)

    return np.array(images)


X_train = load_images(X_train_paths)
X_val = load_images(X_val_paths)
X_test = load_images(X_test_paths)

print("Training image shape:", X_train.shape)
print("Validation image shape:", X_val.shape)
print("Testing image shape:", X_test.shape)
```

```python
def create_cnn(
    filters=32,
    learning_rate=0.001,
    dropout=0.5
):

    model = models.Sequential([

        layers.Input(shape=(128, 128, 1)),

        layers.Conv2D(
            filters,
            (3, 3),
            activation="relu"
        ),

        layers.MaxPooling2D((2, 2)),

        layers.Conv2D(
            filters * 2,
            (3, 3),
            activation="relu"
        ),

        layers.MaxPooling2D((2, 2)),

        layers.Conv2D(
            filters * 4,
            (3, 3),
            activation="relu"
        ),

        layers.MaxPooling2D((2, 2)),

        layers.Flatten(),

        layers.Dense(128, activation="relu"),

        layers.Dropout(dropout),

        layers.Dense(1, activation="sigmoid")
    ])

    optimizer = tf.keras.optimizers.Adam(
        learning_rate=learning_rate
    )

    model.compile(
        optimizer=optimizer,
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

```python
cnn_model = create_cnn()

cnn_model.summary()
```

```python
history = cnn_model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=10,
    batch_size=32,
    verbose=1
)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("CNN Training vs Validation Accuracy")
plt.legend()
plt.show()
```

```python
from sklearn.model_selection import StratifiedKFold

cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

parameter_sets = [
    {
        "filters": 16,
        "learning_rate": 0.001,
        "dropout": 0.5
    },
    {
        "filters": 32,
        "learning_rate": 0.001,
        "dropout": 0.5
    },
    {
        "filters": 32,
        "learning_rate": 0.0001,
        "dropout": 0.5
    }
]

print("Number of parameter combinations:", len(parameter_sets))
```

```python
CV_SAMPLES = min(1500, len(X_train))

X_cv = X_train[:CV_SAMPLES]
y_cv = y_train[:CV_SAMPLES]

cv_results = []

for params in parameter_sets:

    fold_scores = []

    print("\nTesting:", params)

    for fold, (train_idx, val_idx) in enumerate(
        cv.split(X_cv, y_cv), start=1
    ):

        model = create_cnn(
            filters=params["filters"],
            learning_rate=params["learning_rate"],
            dropout=params["dropout"]
        )

        model.fit(
            X_cv[train_idx],
            y_cv[train_idx],
            validation_data=(
                X_cv[val_idx],
                y_cv[val_idx]
            ),
            epochs=3,
            batch_size=32,
            verbose=0
        )

        predictions = (
            model.predict(
                X_cv[val_idx],
                verbose=0
            ).ravel() >= 0.5
        ).astype(int)

        score = accuracy_score(
            y_cv[val_idx],
            predictions
        )

        fold_scores.append(score)

        print(
            "Fold",
            fold,
            "Accuracy:",
            round(score, 4)
        )

    mean_score = np.mean(fold_scores)

    cv_results.append({
        "Parameters": params,
        "Mean CV Accuracy": mean_score
    })

    print(
        "Mean CV Accuracy:",
        round(mean_score, 4)
    )
```

```python
cv_results_df = pd.DataFrame(cv_results)

cv_results_df
```

```python
best_index = cv_results_df["Mean CV Accuracy"].idxmax()

best_parameters = cv_results_df.loc[
    best_index,
    "Parameters"
]

print("Selected hyperparameters:")
print(best_parameters)
```

```python
final_model = create_cnn(
    filters=best_parameters["filters"],
    learning_rate=best_parameters["learning_rate"],
    dropout=best_parameters["dropout"]
)

final_history = final_model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=10,
    batch_size=32,
    verbose=1
)
```

```python
y_probability = final_model.predict(
    X_test,
    verbose=1
).ravel()

y_prediction = (
    y_probability >= 0.5
).astype(int)

print("Predictions generated.")
```
